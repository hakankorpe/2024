```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        ltr = 0
        rtr = len(s) - 1
        s= s.lower()
        while ltr < rtr:
            if s[ltr].isalnum() and s[rtr].isalnum():
                if s[ltr] == s[rtr]:
                    ltr += 1
                    rtr -= 1
                else:
                    return False
            elif s[ltr].isalnum():
                rtr -=1
            else:
                ltr +=1

        return True
```
