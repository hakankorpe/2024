```python
class Solution:
    def isSubsequence(self, s: str, t: str) -> bool:
        s_ptr = 0
        t_ptr = 0

        while t_ptr < len(t):
            if s_ptr >= len(s):
                return True
            elif t[t_ptr] == s[s_ptr]:
                s_ptr +=1
            t_ptr+=1

        if s_ptr < len(s):
            return False
        else:
            return True



```
